# Review recommend vs not

Product review text → binary recommend label → AutoKeras classifier.

Place the Kaggle **Women’s Clothing E-Commerce Reviews** CSV at `data/review.csv`
(columns used: `text`, `label` where 1 = recommend).


```python
# %pip install autokeras tensorflow scikit-learn pandas numpy

import numpy as np
import pandas as pd
import autokeras as ak
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
# Local file — no Colab upload required
df = pd.read_csv("data/review.csv")
# Expect columns: text, label (1 = recommend)
assert {"text", "label"}.issubset(df.columns)
print(df["label"].value_counts())
df.sample(n=5, random_state=222)
```

```python
x = np.array(df["text"].astype(str))
y = np.array(df["label"].astype(int))

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=222, stratify=y
)
```

```python
clf = ak.TextClassifier(max_trials=2)
clf.fit(x_train, y_train, validation_split=0.3)
```

```python
y_hat = clf.predict(x_test)
print(classification_report(y_test, y_hat))
```

```python
text = np.array([
    "Wanted to love this top, but couldn't. the arm holes where so big you have to wear a top under..."
])
pred = clf.predict(text)
print("likely recommend" if int(pred[0]) > 0 else "likely do not recommend")
```
